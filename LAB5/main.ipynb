Exp-5.1 

(a) Using Pretrained Model

```python
from transformers import GPT2Tokenizer

# Load pretrained BPE tokenizer
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")

# Read input data
with open("input_sub_word_data.txt", "r", encoding="utf-8") as f:
    text = f.read()

# Tokenize
tokens = tokenizer.tokenize(text)

# Convert tokens to IDs
token_ids = tokenizer.convert_tokens_to_ids(tokens)

# Display
print("Tokens:")
print(tokens)

print("\nToken IDs:")
print(token_ids)

print("\nToken - ID:")
for token, token_id in zip(tokens, token_ids):
    print(f"{token:20} -> {token_id}")
```

(b) Without using Pretrained Model

```python
%pip install tokenizers
```

```python
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace

# Read corpus
with open("input_sub_word_data.txt", "r", encoding="utf-8") as f:
    text = f.read()

# Create BPE tokenizer
tokenizer = Tokenizer(BPE(unk_token="[UNK]"))

# Split text into words
tokenizer.pre_tokenizer = Whitespace()

# BPE trainer
trainer = BpeTrainer(
    vocab_size=500,
    min_frequency=2,
    special_tokens=["[UNK]", "[PAD]"]
)

# Train tokenizer
tokenizer.train_from_iterator([text], trainer)

# Tokenize
output = tokenizer.encode(text)

print("Tokens:")
print(output.tokens)

print("\nToken IDs:")
print(output.ids)

print("\nToken - ID:")
for token, token_id in zip(output.tokens, output.ids):
    print(f"{token:20} -> {token_id}")
```

Experiment 5.2

(b) Without using pretrained model

```python
import sentencepiece as spm

input_file = "input_sub_word_data.txt"
model_prefix = "custom_sentencepiece"

# Train SentencePiece tokenizer
spm.SentencePieceTrainer.train(
    input=input_file,
    model_prefix=model_prefix,
    vocab_size=300,
    model_type="bpe",
    character_coverage=1.0
)

# Load trained model
sp = spm.SentencePieceProcessor(
    model_file=model_prefix + ".model"
)

# Read input
with open(input_file, "r", encoding="utf-8") as f:
    text = f.read()

# Tokenize
tokens = sp.encode(text, out_type=str)

# Token IDs
token_ids = sp.encode(text, out_type=int)

print("Tokens:")
print(tokens)

print("\nToken IDs:")
print(token_ids)

print("\nToken - ID:")
for token, token_id in zip(tokens, token_ids):
    print(f"{token:20} -> {token_id}")
```

Experiment 5.3
 Implement Part-of-Speech (POS) tagging on a given text and display the unique token, POS tag, and description of each POS tag.

POS tagging

(a) using Spacy

```python
import spacy

# Load English model
nlp = spacy.load("en_core_web_sm")

text = "The young student is reading an interesting book in the library."

doc = nlp(text)

print(f"{'Token':15} {'POS':10} {'Description'}")
print("-" * 50)

for token in doc:
    print(f"{token.text:15} {token.pos_:10} {spacy.explain(token.pos_)}")
```

```python
%pip install nltk
```

```python
import nltk

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger_eng")
nltk.download("universal_tagset")
```

```python
import nltk
from nltk import word_tokenize, pos_tag
from nltk.tag import map_tag

text = "The young student is reading an interesting book in the library."

# Tokenize
tokens = word_tokenize(text)

# POS tagging
tags = pos_tag(tokens)

print(f"{'Token':15} {'POS':10} {'Description'}")
print("-" * 50)

for token, tag in tags:
    description = map_tag("en-ptb", "universal", tag)
    print(f"{token:15} {tag:10} {description}")
```

Experiment 5.4

Implement Part-of-Speech (POS) tagging on a given text and display the unique token, POS tag, description of each token and frequency using Spacy.

Input data: Same input data of 5.1

```python
import spacy
from collections import Counter

# Load English model
nlp = spacy.load("en_core_web_sm")

# Read input data
with open("input_sub_word_data.txt", "r", encoding="utf-8") as f:
    text = f.read()

# Process text
doc = nlp(text)

# Filter out whitespace tokens
tokens = [token for token in doc if not token.is_space]

# Calculate token frequency (case-insensitive for accurate count)
token_counts = Counter(token.text.lower() for token in tokens)

# Track unique tokens with POS tag, description, and frequency
unique_tokens = {}
for token in tokens:
    key = token.text.lower()
    if key not in unique_tokens:
        unique_tokens[key] = {
            "pos": token.pos_,
            "description": spacy.explain(token.pos_),
            "frequency": token_counts[key]
        }

# Display unique tokens, POS tag, description, and frequency
print(f"{'Token':20} {'POS':10} {'Description':25} {'Frequency'}")
print("-" * 65)

for token, info in unique_tokens.items():
    print(f"{token:20} {info['pos']:10} {str(info['description']):25} {info['frequency']}")

print()
print("=" * 65)
print(f"Total tokens: {len(tokens)}")
print(f"Total unique tokens: {len(unique_tokens)}")
print(f"Sum of frequencies: {sum(token_counts.values())}")
```

Experiment 5.5

Write a Python program to train a Word2Vec CBOW model using the given text corpus.
Display the vector representation of a selected word and find the three most similar words.

Input corpus: Create a file named corpus.txt

```python
%pip install gensim
```

```python
from gensim.models import Word2Vec

# Read and preprocess the corpus
with open("corpus.txt", "r", encoding="utf-8") as f:
    sentences = [line.strip().lower().split() for line in f if line.strip()]

print(f"Preprocessed {len(sentences)} sentences from corpus.txt:")
for i, sent in enumerate(sentences, 1):
    print(f"  {i}. {sent}")

# Train Word2Vec CBOW model (sg=0 indicates Continuous Bag-of-Words)
cbow_model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=3,
    min_count=1,
    sg=0,
    seed=42,
    epochs=200
)

# Select a word from vocabulary
selected_word = "machine"

# Display vector representation
print()
print(f"Vector representation of '{selected_word}':")
print(cbow_model.wv[selected_word])
print(f"Vector shape: {cbow_model.wv[selected_word].shape}")

# Find the three most similar words
print()
print(f"Three most similar words to '{selected_word}':")
similar_words = cbow_model.wv.most_similar(selected_word, topn=3)
for word, similarity in similar_words:
    print(f"  {word:15} -> Cosine Similarity: {similarity:.4f}")
```

Experiment 5.6

Write a Python program to train a Skip-Gram Word2Vec model using the same corpus.
Retrieve the vector for "language" and display the five most similar words.

Input corpus: Use corpus.txt from Question 5.5.

```python
from gensim.models import Word2Vec

# Read and preprocess the corpus
with open("corpus.txt", "r", encoding="utf-8") as f:
    sentences = [line.strip().lower().split() for line in f if line.strip()]

# Train Word2Vec Skip-Gram model (sg=1 indicates Skip-Gram)
skipgram_model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=3,
    min_count=1,
    sg=1,
    seed=42,
    epochs=200
)

# Retrieve vector for "language"
target_word = "language"
print(f"Vector representation of '{target_word}':")
print(skipgram_model.wv[target_word])
print(f"Vector shape: {skipgram_model.wv[target_word].shape}")

# Display the five most similar words
print()
print(f"Five most similar words to '{target_word}':")
similar_words = skipgram_model.wv.most_similar(target_word, topn=5)
for word, similarity in similar_words:
    print(f"  {word:15} -> Cosine Similarity: {similarity:.4f}")
```
