Experiment-1.1: Read the employee dataset and display the average salary of each department using a bar chart.

```python
# data loading
import pandas as pd
df = pd.read_csv('/Users/himanshu/Desktop/coding/NLP Lab/employee_information_100.csv')
df.info()
```

```python
average_val = df['Salary'].mean()
print(average_val)
```

Experiment 1.2: Display the number of employees working in each department.

```python
number_of_employee = df.groupby('Department').size()
print(number_of_employee)
```

```python

df['Department'].value_counts().plot(kind='bar', title='Employees per Department')
```

Experiment 1.3:Display the percentage of male and female employees using a pie chart

```python
import matplotlib.pyplot as plt

gender_cnt = df['Gender'].value_counts()

gender_cnt.plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90,
    ylabel=''
)
plt.title("Employee based on Gender")
plt.show()
```

Experiment 1.4: Draw a histogram showing salary distribution.

```python
plt.figure(figsize=(6.4,4.8));
plt.hist(
    df['Salary'],
    bins=15,
    color='green',
    edgecolor='black'
)
plt.title("Employee Salary Distribution",fontsize=14, fontweight='bold', pad = 10)
plt.xlabel("Salary",fontsize=12)
plt.ylabel("Frequency",fontsize=12)
```

Experiment 1.5: Draw a scatter plot showing the relationship between experience and salary.

```python
plt.figure(figsize=(6.4,4.8));
plt.scatter(
    df['Experience_Years'],
    df['Salary'],
    color='blue',
    edgecolor='black'
)
plt.title("Experience vs Salary",fontsize=14, fontweight='bold', pad = 10)
plt.xlabel("Experience",fontsize=12)
plt.ylabel("Salary",fontsize=12)
plt.show()
```

Experiment 1.6: Display the top 10 employees with the highest salary.

```python
top10_values=df.sort_values(by='Salary',ascending=False).head(10)
print(top10_values[['Name','Salary','Department']])
```

Experiment 1.7: Find the highest salary in every department.

```python
highest_val = df.groupby('Department')['Salary'].max()
print(highest_val)
```

Experiment 1.8: Display employees whose salary is greater than the overall average salary.

```python
avg_sal=df['Salary'].mean()
above_avg=df[df['Salary']>avg_sal]
print(above_avg[['Name','Salary','Department']])
```

Experiment 1.9: Calculate the average years of experience in each department.

```python
avg_yr_exp = df.groupby('Department')['Experience_Years'].mean()
print(avg_yr_exp)
```

Experiment 1.10: Display the distribution of employee ages using a histogram.

```python
plt.figure(figsize=(6.4,4.8));
plt.hist(
    df['Age'],
    bins=15,
    color='green',
    edgecolor='black'
)
plt.title("Employee Age Distribution",fontsize=14, fontweight='bold', pad = 10)
plt.xlabel("Age",fontsize=12)
plt.ylabel("Frequency",fontsize=12)
```
