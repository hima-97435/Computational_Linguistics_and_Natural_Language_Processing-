Experiment 3.1: 

 Write a Python program to perform  Write a Python program to perform stemming using the Porter Stemmer. Apply stemming to the following words. Display the original words and their stems.
playing, played, plays, studies, studying, connected, connection, computers


```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
# It removes common suffixes from English words to reduce them to their root base forms,
# making it a fundamental text normalization tool in information retrieval and search engines
words = [
    "playing",
    "played",
    "plays",
    "studies",
    "studying",
    "connected",
    "connection",
    "computers"
]

print("Original Word \t Stem")
print("-"*30)

for word in words:
    stem = stemmer.stem(word)
    print(f"{word}\t\t{stem}")
```

Experiment 3.2:  Write a Python program to perform lemmatization on a given input words:

cats,	dogs,   running,  runs,	ran, studies, studying,	better, children, mice,	went, ate,	
leaves,	caring,


```python
import nltk
from nltk.stem import WordNetLemmatizer

nltk.download("wordnet")
nltk.download("omw-1.4")

lemmatizer=WordNetLemmatizer()
words = [
    "cats",
    "dogs",
    "running",
    "runs",
    "ran",
    "studies",
    "studying",
    "better",
    "children",
    "mice",
    "went",
    "ate",
    "leaves",
    "caring"
]
pos_tags = {
    "cats": "n",
    "dogs": "n",
    "running": "v",
    "runs": "v",
    "ran": "v",
    "studies": "v",
    "studying": "v",
    "better": "a",
    "children": "n",
    "mice": "n",
    "went": "v",
    "ate": "v",
    "leaves": "n",
    "caring": "v"
}
print("\nOriginal Word \t Lemma")
print("-"*30)

for word in words:
    pos = pos_tags[word]
    lemma = lemmatizer.lemmatize(word, pos=pos)
    print(f"{word}\t\t{lemma}")
```

```python
# Experiment 3.3
# Extract Email, URLs, Mobile Numbers, Hashtags and Mentions using Regex

import re

text = """
Welcome to the Natural Language Processing workshop! For more information,
visit https://www.nlpworkshop.com or www.python.org.
You can contact the coordinator at nlpworkshop@gmail.com or support@python.org.
For registration, call +91-9876543210 or 9123456789.
Follow us on social media @NLPWorkshop and @PythonLearner.
Share your experience using #NLP, #Python, and #MachineLearning.
You can also visit https://github.com/NLPWorkshop for the latest updates.
"""

# 1. Extract Email Addresses
emails = re.findall(
    r'[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}',
    text
)

# 2. Extract URLs
urls = re.findall(
    r'(?:https?://|www\.)[A-Za-z0-9.-]+\.[A-Za-z]{2,}(?:/[A-Za-z0-9._~:/?#\[\]@!$&\'()*+,;=%-]*)?',
    text
)

# 3. Extract Mobile Numbers
mobile_numbers = re.findall(
    r'(?:\+91[-\s]?)?[6-9]\d{9}',
    text
)

# 4. Extract Hashtags
hashtags = re.findall(
    r'#[A-Za-z0-9_]+',
    text
)

# 5. Extract Mentions
mentions = re.findall(
    r'@[A-Za-z0-9_]+',
    text
)

# Display results

print("EMAIL ADDRESSES:")
for email in emails:
    print(email)

print("\nURLS:")
for url in urls:
    print(url)

print("\nMOBILE NUMBERS:")
for number in mobile_numbers:
    print(number)

print("\nHASHTAGS:")
for hashtag in hashtags:
    print(hashtag)

print("\nMENTIONS:")
for mention in mentions:
    print(mention)
```
