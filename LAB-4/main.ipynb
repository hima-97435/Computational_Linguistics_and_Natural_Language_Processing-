4.1. To perform Term-Frequency Analysis and Named Entity Recognition on a given text using any toolkit.

Calculate the frequency of each word and save the output in CSV format (Term, Frequency)

Identify the top 10 most frequent terms and display their corresponding frequencies.


```python


import nltk
import spacy
import csv
from collections import Counter

# nltk.download("punkt")

# Load NLP model
nlp = spacy.load("en_core_web_sm")

# -------------------------------
# 1. Term Frequency Analysis
# -------------------------------

# Tokenize words using NLTK
tokens = nltk.word_tokenize(text)

# Keep only alphabetic words
words = [word.lower() for word in tokens if word.isalpha()]

# Calculate frequency
frequency = Counter(words)

# Save frequency to CSV
with open("term_frequency_toolkit.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    # Header
    writer.writerow(["Term", "Frequency"])

    # Write terms and frequencies
    for term, freq in frequency.most_common():
        writer.writerow([term, freq])

# Display top 10
print("Top 10 Most Frequent Terms:")
for term, freq in frequency.most_common(10):
    print(term, ":", freq)


# -------------------------------
# 2. Named Entity Recognition
# -------------------------------

doc = nlp(text)

print("\nNamed Entities:")

for entity in doc.ents:
    print(entity.text, "->", entity.label_)

# ORG --> Organization
# GPE --> Geopolitical Entity
# PERSON --> Person
# LOC --> Location
# DATE --> Date
```

4.2. To perform Term-Frequency Analysis on a given text without using any toolkit.

Calculate the frequency of each word and save the output in CSV format (Term, Frequency)

Identify the top 10 most frequent terms and display their corresponding frequencies.


```python
import string

with open("4.1_4.2_input.txt","r" , encoding ="utf-8") as file:
    text=file.read()
```

```python
import csv
import re



# --------------------------------
# 1. Convert text to lowercase
# --------------------------------

text2 = text.lower()

# --------------------------------
# 2. Extract words manually
# --------------------------------

words = re.findall(r'\b[a-z]+\b', text2)

# --------------------------------
# 3. Calculate frequency manually
# --------------------------------

frequency = {}

for word in words:

    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

# --------------------------------
# 4. Sort by frequency
# --------------------------------

sorted_frequency = sorted(
    frequency.items(),
    key=lambda x: x[1],
    reverse=True
)

# --------------------------------
# 5. Save to CSV
# --------------------------------

with open("term_frequency_no_toolkit.csv", "w", newline="", encoding="utf-8") as file:

    writer = csv.writer(file)

    writer.writerow(["Term", "Frequency"])

    for term, freq in sorted_frequency:
        writer.writerow([term, freq])

# --------------------------------
# 6. Display top 10
# --------------------------------

print("Top 10 Most Frequent Terms:")

for term, freq in sorted_frequency[:10]:
    print(term, ":", freq)
```

## 4.3 TF-IDF Analysis Without NLP Toolkits

Write a Python program **without using NLTK, spaCy, or any other NLP toolkit** to perform the following tasks:

1. **Read Multiple Text Documents**
   - Read and process multiple text documents as input.

2. **Convert Text to Lowercase**
   - Convert all text to lowercase to ensure consistent term matching.

3. **Remove Punctuation**
   - Remove punctuation marks from the text.

4. **Tokenize the Documents**
   - Split each document into individual terms/words.

5. **Calculate Term Frequency (TF)**
   - Calculate the **TF** of every term in each document.

6. **Calculate Document Frequency (DF)**
   - Calculate the **DF** of every term across all documents.

7. **Calculate Inverse Document Frequency (IDF)**
   - Calculate **IDF** using the logarithmic formula:
   
   \[
   IDF(t) = \log\left(\frac{N}{DF(t)}\right)
   \]
   
   where:
   - `N` = total number of documents
   - `DF(t)` = number of documents containing term `t`

8. **Calculate TF-IDF**
   - Calculate the **TF-IDF** score for every term in every document using:
   
   \[
   TF\text{-}IDF(t,d) = TF(t,d) \times IDF(t)
   \]

9. **Display TF, DF, IDF, and TF-IDF**
   - Display the following information for every term:
   
   | Term | TF | DF | IDF | TF-IDF |
   |------|----|----|-----|--------|

10. **Identify Top 10 Terms**
    - Identify and display the **top 10 terms with the highest TF-IDF scores** for each document.

```python
import math
import string
from collections import defaultdict


# --------------------------------------------------
# INPUT DOCUMENTS
# --------------------------------------------------

documents = {
    "Document 1": "Natural language processing is a field of artificial intelligence.",
    "Document 2": "Natural language processing helps computers understand human language.",
    "Document 3": "Machine learning is an important part of artificial intelligence."
}


# --------------------------------------------------
# STEP 1-4: LOWERCASE, REMOVE PUNCTUATION, TOKENIZE
# --------------------------------------------------

tokenized_documents = {}

for doc_name, text in documents.items():

    # Convert to lowercase
    text = text.lower()

    # Remove punctuation
    text = text.translate(str.maketrans("", "", string.punctuation))

    # Tokenize
    words = text.split()

    tokenized_documents[doc_name] = words

    print("\n", doc_name)
    print("Tokens:", words)


# --------------------------------------------------
# STEP 5: CALCULATE TF
# --------------------------------------------------

tf = {}

for doc_name, words in tokenized_documents.items():

    tf[doc_name] = {}

    total_words = len(words)

    # Count each word
    word_count = {}

    for word in words:
        if word in word_count:
            word_count[word] += 1
        else:
            word_count[word] = 1

    # TF = frequency / total words
    for word, count in word_count.items():
        tf[doc_name][word] = count / total_words


# --------------------------------------------------
# STEP 6: CALCULATE DF
# --------------------------------------------------

df = {}

for doc_name, words in tokenized_documents.items():

    # set() ensures a word is counted only once per document
    unique_words = set(words)

    for word in unique_words:

        if word in df:
            df[word] += 1
        else:
            df[word] = 1


# --------------------------------------------------
# STEP 7: CALCULATE IDF
# --------------------------------------------------

N = len(documents)

idf = {}

for word, document_frequency in df.items():

    idf[word] = math.log(N / document_frequency)


# --------------------------------------------------
# STEP 8: CALCULATE TF-IDF
# --------------------------------------------------

tfidf = {}

for doc_name in documents:

    tfidf[doc_name] = {}

    for word in tf[doc_name]:

        tfidf[doc_name][word] = (
            tf[doc_name][word] * idf[word]
        )


# --------------------------------------------------
# STEP 9: DISPLAY TERM, TF, DF, IDF, TF-IDF
# --------------------------------------------------

print("\n\n==============================================")
print("TF - DF - IDF - TF-IDF")
print("==============================================")

for doc_name in documents:

    print("\n" + doc_name)
    print("-" * 65)

    print(
        f"{'Term':<20}"
        f"{'TF':<12}"
        f"{'DF':<8}"
        f"{'IDF':<12}"
        f"{'TF-IDF':<12}"
    )

    print("-" * 65)

    for word in tf[doc_name]:

        print(
            f"{word:<20}"
            f"{tf[doc_name][word]:<12.4f}"
            f"{df[word]:<8}"
            f"{idf[word]:<12.4f}"
            f"{tfidf[doc_name][word]:<12.4f}"
        )


# --------------------------------------------------
# STEP 10: TOP 10 TERMS FOR EACH DOCUMENT
# --------------------------------------------------

print("\n\n==============================================")
print("TOP 10 TERMS BY TF-IDF")
print("==============================================")

for doc_name in documents:

    # Sort terms according to TF-IDF
    sorted_terms = sorted(
        tfidf[doc_name].items(),
        key=lambda x: x[1],
        reverse=True
    )

    print("\n" + doc_name)
    print("-" * 40)

    for term, score in sorted_terms[:10]:

        print(f"{term:<20} {score:.4f}")
```
