Experiment 2.1:  Write a Python program to perform basic text preprocessing on a given data. The program should:\\

1. Convert text to lowercase. (Also display how many uppercase?)

2. Remove punctuation.        (Also display how many different types of punctuations?)

3. Remove numbers.            (Also display how many numbers?)

4. Remove extra whitespace. (How much extra whitespace is present?)

5. Display the cleaned text.     


```python
import string


with open("2.1_text_data.txt","r" , encoding ="utf-8") as file:
    text=file.read()
#0 display the original text
# print(text)

#1 lowercase text
upp_cnt = sum(1 for c in text if c.isupper())
print(f"Number of uppercase letters: {upp_cnt}")

text=text.lower()
# print(text)

#2 remove punctuation
punc_cnt = sum(1 for c in text if c in string.punctuation)
print(f"Number of punctuation characters: {punc_cnt}")
text="".join([c for c in text if c not in string.punctuation])
# print(text)

#3 remove numbers
num_cnt = sum(1 for c in text if c.isdigit())
print(f"Number of digits: {num_cnt}")
text= "".join([c for c in text if not c.isdigit()])
# print(text)

#4 remove extra space
space_cnt = sum(1 for c in text if c == " ")
print(f"Number of spaces: {space_cnt}")
text="".join(text.split())
# print(text)

#5 display the cleaned text
# print(text)

```

**Using NLTK**

```python
import nltk
nltk.download('punkt')
```

```python
from nltk.tokenize import word_tokenize,sent_tokenize

sentences = sent_tokenize(text)
print(sentences)
print(f"Number of sentences: {len(sentences)}")

words = word_tokenize(text)
print(words)
print(f"Number of words: {len(words)}")
```

```python
import spacy

nlp=spacy.load("en_core_web_sm")
doc=nlp(text)

sentences=[sent.text for sent in doc.sents]
print(f"Number of sentences: {len(sentences)}")
for idx,sentences in enumerate(sentences,1):
    print(f"Sentence {idx}: {sentences}")

words=[token.text for token in doc]
print(f"Number of words: {len(words)}")
for idx,words in enumerate(words,1):
    print(f"Word {idx}: {words}")
```

**without using nltk/spacy**

```python
import re
sentence_patterns = r'(?<=[.!?])\s+'
sentences = re.split(sentence_patterns, text)
sentences

word_patterns = r'\w+|[^\w\s]'
words =re.findall(word_patterns, text)



print(sentences)

print(words)
```

Experiment 2.3: Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens. (Also, display Stop Words Separately).

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

nltk.download('stopwords')
nltk.download('punkt')


```

```python
with open("2.3_clean_data.txt","r",encoding="utf-8") as file:
    text=file.read()

tokens= word_tokenize(text)

stop_words=set(stopwords.words("english"))

print("Original Tokens:")
print(tokens)

found_stopwords=[token for token in tokens if token.lower() in stop_words]
print("\nFound Stopwords:")
print(found_stopwords)

filtered_tokens=[token for token in tokens if token.lower() not in stop_words]
print("\nFiltered Tokens:")
print(filtered_tokens)
```
